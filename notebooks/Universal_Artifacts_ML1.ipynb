# Universal ECG artifact-removal model (ML1)

This notebook trains the existing residual CNN on many MIT-BIH records, with a record-wise validation split. It creates an even mix of noisy and clean examples on demand, so the model learns both artifact removal and when to remove nothing.

The supervised target is `DSP(noisy ECG) - DSP(clean ECG)`. Therefore `recovered = input - predicted_artifact` is compared with the filtered clean ECG, without mixing filtered and unfiltered signals.

```python
def bandpass_filter(signal, fs=FS):
    """0.5--40 Hz zero-phase Butterworth filter used by ML1."""
    sos = butter(2, [0.5, 40], btype='bandpass', fs=fs, output='sos')
    return sosfiltfilt(sos, np.asarray(signal, dtype=np.float64))

def read_lead(record_name, database):
    local_record = DATA_ROOT / database / record_name
    # Prefer the explicit local download; retain the PhysioNet fallback for convenience.
    # NSTDB must be read as digital values: its header has physical gain 0.
    read_options = {'physical': database != 'nstdb'}
    record = (wfdb.rdrecord(str(local_record), **read_options) if local_record.with_suffix('.hea').exists()
              else wfdb.rdrecord(record_name, pn_dir=database, **read_options))
    if round(record.fs) != FS:
        raise ValueError(f'{database}/{record_name} has fs={record.fs}; expected {FS} Hz')
    if database == 'nstdb':
        # NSTDB headers use ADC gain 0. p_signal therefore contains inf/NaN; use
        # finite digital samples and normalize each source to unit RMS instead.
        signal = np.asarray(record.d_signal[:, 0], dtype=np.float64)
        signal = signal[np.isfinite(signal)] if not np.isfinite(signal).all() else signal
        signal = signal - np.median(signal)
        rms = np.sqrt(np.mean(signal**2))
        if not np.isfinite(rms) or rms <= 0: raise ValueError(f'Invalid NSTDB noise record: {record_name}')
        return signal / rms
    signal = np.asarray(record.p_signal[:, 0], dtype=np.float64)
    if not np.isfinite(signal).all(): raise ValueError(f'Non-finite samples in {database}/{record_name}')
    return signal

def load_sources(record_names, database):
    sources = {}
    for name in record_names:
        print(f'Loading {database}/{name} ...')
        sources[name] = read_lead(name, database)
    return sources

train_raw = load_sources(TRAIN_RECORDS, 'mitdb')
val_raw = load_sources(VAL_RECORDS, 'mitdb')
noise_raw = load_sources(NOISE_RECORDS, 'nstdb')
print({name: signal.shape for name, signal in train_raw.items()})
```

```python
class ArtifactWindowDataset(Dataset):
    """Deterministically samples clean/noisy windows from whole ECG records.

    Synthetic artifact type, position, amplitude, and clean/noisy choice change each
    training epoch. Validation is deterministic and uses held-out records only.
    """
    def __init__(self, records, noises, windows_per_record, scale, seed, training):
        self.records = records
        self.noises = list(noises.values())
        self.scale = float(scale)
        self.seed = seed
        self.training = training
        self.epoch = 0
        self.items = []
        rng = np.random.default_rng(seed)
        for record_name, raw in records.items():
            high = len(raw) - WINDOW - CONTEXT
            if high <= CONTEXT:
                raise ValueError(f'Record {record_name} is too short for a {WINDOW}-sample window')
            starts = rng.integers(CONTEXT, high, size=windows_per_record)
            self.items.extend((record_name, int(start)) for start in starts)

    def set_epoch(self, epoch):
        self.epoch = epoch

    def __len__(self):
        return len(self.items)

    def __getitem__(self, index):
        record_name, start = self.items[index]
        raw = self.records[record_name]
        # A reproducible per-example RNG means validation is stable and reruns are comparable.
        epoch_key = self.epoch if self.training else 0
        rng = np.random.default_rng(np.random.SeedSequence([self.seed, epoch_key, index]))
        left, right = start - CONTEXT, start + WINDOW + CONTEXT
        clean_segment = raw[left:right]
        filtered_clean = bandpass_filter(clean_segment)[CONTEXT:CONTEXT + WINDOW]

        if rng.random() < CLEAN_FRACTION:
            model_input = filtered_clean
            target = np.zeros(WINDOW, dtype=np.float64)
        else:
            # Include single-source and mixed NSTDB artifacts. Every noisy example gets
            # one target input SNR, so the model sees almost-clean through severe ECG.
            count = int(rng.choice([1, 2, 3], p=[0.50, 0.35, 0.15]))
            selected = rng.choice(len(self.noises), size=count, replace=False)
            artifact = np.zeros_like(clean_segment)
            for noise_id in selected:
                noise = self.noises[int(noise_id)]
                noise_start = int(rng.integers(0, len(noise) - len(clean_segment)))
                artifact += noise[noise_start:noise_start + len(clean_segment)]
            target_snr_db = float(rng.choice([-6, -3, 0, 3, 6, 9, 12, 15, 20]))
            clean_rms = np.sqrt(np.mean(clean_segment ** 2))
            artifact_rms = np.sqrt(np.mean(artifact ** 2))
            if clean_rms <= 0 or artifact_rms <= 0:
                raise ValueError('Cannot create a finite-SNR example from zero-RMS data')
            artifact *= clean_rms / (artifact_rms * 10 ** (target_snr_db / 20))
            model_input = bandpass_filter(clean_segment + artifact)[CONTEXT:CONTEXT + WINDOW]
            target = model_input - filtered_clean

        x = torch.from_numpy((model_input / self.scale).astype(np.float32)).unsqueeze(0)
        y = torch.from_numpy((target / self.scale).astype(np.float32)).unsqueeze(0)
        return x, y

# Rebuild NSTDB noise here as well. This makes the dataset cell safe to re-run even
# if an earlier notebook cell left an old p_signal-based (NaN) noise_raw in memory.
def load_normalized_nstdb_noise(record_name):
    local_record = DATA_ROOT / 'nstdb' / record_name
    record = (wfdb.rdrecord(str(local_record), physical=False) if local_record.with_suffix('.hea').exists()
              else wfdb.rdrecord(record_name, pn_dir='nstdb', physical=False))
    signal = np.asarray(record.d_signal[:, 0], dtype=np.float64)
    if not np.isfinite(signal).all():
        raise ValueError(f'NSTDB digital record {record_name} contains non-finite samples')
    signal = signal - np.median(signal)
    rms = np.sqrt(np.mean(signal**2))
    if rms <= 0 or not np.isfinite(rms): raise ValueError(f'Invalid NSTDB noise record: {record_name}')
    return signal / rms

noise_raw = {name: load_normalized_nstdb_noise(name) for name in NOISE_RECORDS}
assert all(np.isfinite(signal).all() for signal in noise_raw.values())
print('NSTDB noise reloaded as finite, unit-RMS digital signals.')

# One fixed scale preserves signal-to-artifact amplitudes, unlike independent per-window normalization.
scale_samples = np.concatenate([signal[::360] for signal in train_raw.values()])
NORMALIZATION_SCALE = float(np.percentile(np.abs(scale_samples), 99.5))
if NORMALIZATION_SCALE <= 0:
    raise ValueError('Invalid normalization scale')
train_dataset = ArtifactWindowDataset(train_raw, noise_raw, TRAIN_WINDOWS_PER_RECORD, NORMALIZATION_SCALE, SEED, training=True)
val_dataset = ArtifactWindowDataset(val_raw, noise_raw, VAL_WINDOWS_PER_RECORD, NORMALIZATION_SCALE, SEED + 1, training=False)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=torch.cuda.is_available())
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=torch.cuda.is_available())
probe_x, probe_y = next(iter(train_loader))
assert torch.isfinite(probe_x).all() and torch.isfinite(probe_y).all(), 'Dataset probe is non-finite'
print(f'Normalization scale: {NORMALIZATION_SCALE:.5f}; train windows: {len(train_dataset):,}; validation windows: {len(val_dataset):,}; finite batch probe: OK')
```

```python
class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.conv1 = nn.Conv1d(channels, channels, kernel_size=5, padding=2)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv1d(channels, channels, kernel_size=5, padding=2)

    def forward(self, x):
        return self.relu(self.conv2(self.relu(self.conv1(x))) + x)

class ResidualCNN(nn.Module):
    # Kept intentionally equivalent to the existing ML1 architecture.
    def __init__(self):
        super().__init__()
        self.input_conv = nn.Conv1d(1, 32, kernel_size=5, padding=2)
        self.relu = nn.ReLU()
        self.res1, self.res2, self.res3 = ResidualBlock(32), ResidualBlock(32), ResidualBlock(32)
        self.output_conv = nn.Conv1d(32, 1, kernel_size=5, padding=2)

    def forward(self, x):
        x = self.relu(self.input_conv(x))
        return self.output_conv(self.res3(self.res2(self.res1(x))))

model = ResidualCNN().to(device)
# Huber loss has bounded gradients for rare high-amplitude motion artifacts;
# recovery is still evaluated with MSE/RMSE below.
criterion = nn.HuberLoss(delta=0.5)
# foreach=False avoids multi-tensor optimizer paths that are less reliable on MPS.
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, eps=1e-5, foreach=False)
print(model)
```

```python
def _new_optimizer(lr):
    return torch.optim.AdamW(model.parameters(), lr=lr, eps=1e-5, foreach=False)

def mean_loss(loader, training=False):
    global optimizer
    model.train(training)
    total, count, skipped_updates = 0.0, 0, 0
    context = torch.enable_grad() if training else torch.no_grad()
    with context:
        for batch_index, (x, y) in enumerate(loader):
            # Check on CPU first. MPS finite-reduction has produced false failures on
            # otherwise valid tensors, and DataLoader always gives us CPU tensors here.
            if not (torch.isfinite(x).all().item() and torch.isfinite(y).all().item()):
                raise FloatingPointError(f'Non-finite CPU data in batch {batch_index}; input range={x.nan_to_num().min().item():.3g}..{x.nan_to_num().max().item():.3g}, target range={y.nan_to_num().min().item():.3g}..{y.nan_to_num().max().item():.3g}')
            x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
            predicted_artifact = model(x)
            if not torch.isfinite(predicted_artifact).all():
                raise FloatingPointError('Model produced non-finite predictions; a bad optimizer step was not rolled back.')
            # The former artifact + recovered-ECG losses were algebraically the same MSE twice.
            loss = criterion(predicted_artifact, y)
            if not torch.isfinite(loss):
                raise FloatingPointError('Non-finite loss with finite inputs and predictions.')
            if training:
                optimizer.zero_grad(set_to_none=True)
                loss.backward()
                grad_norm = torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                if not torch.isfinite(grad_norm):
                    skipped_updates += 1
                    optimizer = _new_optimizer(5e-5)
                    continue
                # MPS occasionally corrupts an Adam-family update. Keep a tiny
                # transactional backup and roll back that batch instead of the run.
                backup = [parameter.detach().clone() for parameter in model.parameters()] if device.type == 'mps' else None
                optimizer.step()
                if not all(torch.isfinite(parameter).all() for parameter in model.parameters()):
                    with torch.no_grad():
                        for parameter, previous in zip(model.parameters(), backup): parameter.copy_(previous)
                    skipped_updates += 1
                    current_lr = optimizer.param_groups[0]['lr']
                    optimizer = _new_optimizer(max(current_lr * 0.5, 1e-6))
                    continue
            total += loss.item() * len(x)
            count += len(x)
    if count == 0: raise FloatingPointError('Every optimizer update was rejected as non-finite.')
    if skipped_updates: print(f'Skipped/rolled back {skipped_updates} unstable optimizer updates.')
    return total / count

# Make this training cell self-healing when it is re-run after an earlier NaN run.
# It does not rely on noise_raw/train_loader left behind by previously executed cells.
def _fresh_nstdb_noise_for_training(record_name):
    local_record = DATA_ROOT / 'nstdb' / record_name
    record = (wfdb.rdrecord(str(local_record), physical=False) if local_record.with_suffix('.hea').exists()
              else wfdb.rdrecord(record_name, pn_dir='nstdb', physical=False))
    signal = np.asarray(record.d_signal[:, 0], dtype=np.float64)
    if not np.isfinite(signal).all(): raise FloatingPointError(f'NSTDB {record_name} digital samples are non-finite')
    signal -= np.median(signal)
    rms = np.sqrt(np.mean(signal**2))
    if not np.isfinite(rms) or rms <= 0: raise FloatingPointError(f'NSTDB {record_name} has invalid RMS')
    return signal / rms

fresh_noise = {name: _fresh_nstdb_noise_for_training(name) for name in NOISE_RECORDS}
train_dataset = ArtifactWindowDataset(train_raw, fresh_noise, TRAIN_WINDOWS_PER_RECORD, NORMALIZATION_SCALE, SEED, training=True)
val_dataset = ArtifactWindowDataset(val_raw, fresh_noise, VAL_WINDOWS_PER_RECORD, NORMALIZATION_SCALE, SEED + 1, training=False)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0, pin_memory=torch.cuda.is_available())
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=torch.cuda.is_available())
check_x, check_y = next(iter(train_loader))
if not (torch.isfinite(check_x).all() and torch.isfinite(check_y).all()):
    raise FloatingPointError('Freshly rebuilt dataset is non-finite; training has not started.')
model_device = next(model.parameters()).device
if model_device != device or not all(torch.isfinite(parameter).all() for parameter in model.parameters()):
    print('Resetting model parameters after a device change or earlier non-finite run.')
    model = ResidualCNN().to(device)
    optimizer = _new_optimizer(1e-4)
print('Training refresh complete: finite data and model parameters confirmed.')

best_val, best_state, stale = float('inf'), None, 0
history = {'train': [], 'val': []}
for epoch in range(EPOCHS):
    train_dataset.set_epoch(epoch)
    train_loss = mean_loss(train_loader, training=True)
    val_loss = mean_loss(val_loader, training=False)
    history['train'].append(train_loss); history['val'].append(val_loss)
    print(f'Epoch {epoch + 1:02d} | train Huber loss {train_loss:.6f} | held-out-record val Huber loss {val_loss:.6f}')
    if val_loss < best_val:
        best_val, best_state, stale = val_loss, copy.deepcopy(model.state_dict()), 0
    else:
        stale += 1
        if stale >= PATIENCE:
            print(f'Early stopping after {epoch + 1} epochs.')
            break

model.load_state_dict(best_state)
torch.save({'model_state_dict': model.state_dict(), 'normalization_scale': NORMALIZATION_SCALE, 'fs': FS, 'window': WINDOW, 'stride': STRIDE, 'train_records': TRAIN_RECORDS, 'val_records': VAL_RECORDS, 'best_val_mse': best_val}, CHECKPOINT_PATH)
print(f'Saved best model ({best_val:.6f}) to {CHECKPOINT_PATH.resolve()}')
plt.plot(history['train'], label='train'); plt.plot(history['val'], label='held-out records')
plt.xlabel('Epoch'); plt.ylabel('artifact Huber loss'); plt.legend(); plt.show()
```

```python
def evaluate_recovery(loader):
    mse_before, mse_after, correlations, prds, snr_gains, clean_artifact_rms = [], [], [], [], [], []
    model.eval()
    with torch.no_grad():
        for x, target in loader:
            predicted = model(x.to(device)).cpu()
            x, target = x.numpy()[:, 0], target.numpy()[:, 0]
            predicted = predicted.numpy()[:, 0]
            ideal_clean, recovered = x - target, x - predicted
            error_before, error_after = x - ideal_clean, recovered - ideal_clean
            for clean, before, after, recovered_one, predicted_one in zip(ideal_clean, error_before, error_after, recovered, predicted):
                before_power, after_power = np.mean(before**2), np.mean(after**2)
                signal_power = np.mean(clean**2)
                mse_before.append(before_power); mse_after.append(after_power)
                correlations.append(np.corrcoef(clean, recovered_one)[0, 1])
                prds.append(100 * np.sqrt(np.sum(after**2) / max(np.sum(clean**2), 1e-12)))
                if before_power > 1e-12:
                    snr_gains.append(10 * np.log10(before_power / max(after_power, 1e-12)))
                else:
                    clean_artifact_rms.append(np.sqrt(np.mean(predicted_one**2)))
    return {'mse_before': np.mean(mse_before), 'mse_after': np.mean(mse_after), 'rmse_after': np.sqrt(np.mean(mse_after)), 'correlation': np.nanmean(correlations), 'prd': np.mean(prds), 'snr_improvement_db_noisy_only': np.mean(snr_gains), 'predicted_artifact_rms_clean_only': np.mean(clean_artifact_rms)}

metrics = evaluate_recovery(val_loader)
for name, value in metrics.items(): print(f'{name}: {value:.5f}')
# This is an internal patient-wise result. Repeat the QTDB experiments before declaring ML1 generalizable.
```

## External QTDB validation

Run this only after training. QTDB never enters training or model selection. First test clean QTDB: predicted artifact must stay near zero and ECG shape must stay same. Then add fresh NSTDB noise at every SNR level and check output SNR.


```python
QTDB_RECORDS = ['sel100', 'sel102', 'sel103', 'sel114', 'sel116', 'sel123', 'sel41'] # , 'sel163', 'sel213', 'sel221', 'sel230', 'sel301']
QTDB_WINDOWS_PER_RECORD = 12

def predict_artifact_windows(windows):
    predicted = []
    model.eval()
    with torch.no_grad():
        for first in range(0, len(windows), 256):
            batch = np.asarray(windows[first:first + 256], dtype=np.float32) / NORMALIZATION_SCALE
            output = model(torch.from_numpy(batch).unsqueeze(1).to(device)).cpu().numpy()[:, 0]
            predicted.append(output * NORMALIZATION_SCALE)
    return np.concatenate(predicted)

def qtdb_validation():
    wfdb.dl_database('qtdb', dl_dir=str(DATA_ROOT / 'qtdb'), records=QTDB_RECORDS)
    rng = np.random.default_rng(SEED + 99)
    clean_windows, noisy_windows, clean_targets, snr_labels = [], [], [], []
    for name in QTDB_RECORDS:
        record = wfdb.rdrecord(str(DATA_ROOT / 'qtdb' / name))
        signal = np.asarray(record.p_signal[:, 0], dtype=np.float64)
        if not np.isfinite(signal).all(): raise ValueError(f'QTDB {name} is non-finite')
        signal = resample_poly(signal, FS, round(record.fs)) if round(record.fs) != FS else signal
        starts = rng.integers(CONTEXT, len(signal) - WINDOW - CONTEXT, size=QTDB_WINDOWS_PER_RECORD)
        for start in starts:
            clean_segment = signal[start - CONTEXT:start + WINDOW + CONTEXT]
            clean = bandpass_filter(clean_segment)[CONTEXT:CONTEXT + WINDOW]
            clean_windows.append(clean)
            for requested_snr in [-6, -3, 0, 3, 6, 9, 12, 15, 20]:
                noise = fresh_noise[NOISE_RECORDS[int(rng.integers(len(NOISE_RECORDS)))]]
                noise_start = int(rng.integers(0, len(noise) - len(clean_segment)))
                artifact = noise[noise_start:noise_start + len(clean_segment)]
                artifact *= np.sqrt(np.mean(clean_segment ** 2)) / (np.sqrt(np.mean(artifact ** 2)) * 10 ** (requested_snr / 20))
                noisy = bandpass_filter(clean_segment + artifact)[CONTEXT:CONTEXT + WINDOW]
                noisy_windows.append(noisy); clean_targets.append(clean); snr_labels.append(requested_snr)
    clean_predicted = predict_artifact_windows(clean_windows)
    clean_recovered = np.asarray(clean_windows) - clean_predicted
    clean_metrics = {'rmse': float(np.sqrt(np.mean((clean_recovered - clean_windows) ** 2))), 'correlation': float(np.mean([np.corrcoef(a, b)[0, 1] for a, b in zip(clean_windows, clean_recovered)])), 'predicted_artifact_rms': float(np.sqrt(np.mean(clean_predicted ** 2)))}
    noisy_predicted = predict_artifact_windows(noisy_windows)
    noisy_windows, clean_targets = np.asarray(noisy_windows), np.asarray(clean_targets)
    recovered = noisy_windows - noisy_predicted
    rows = []
    for snr in sorted(set(snr_labels)):
        mask = np.asarray(snr_labels) == snr
        before = noisy_windows[mask] - clean_targets[mask]
        after = recovered[mask] - clean_targets[mask]
        input_snr = 10 * np.log10(np.mean(clean_targets[mask] ** 2) / np.mean(before ** 2))
        output_snr = 10 * np.log10(np.mean(clean_targets[mask] ** 2) / np.mean(after ** 2))
        rows.append({'requested_snr_db': snr, 'input_snr_db': input_snr, 'output_snr_db': output_snr, 'snr_gain_db': output_snr - input_snr, 'mse_before': np.mean(before ** 2), 'mse_after': np.mean(after ** 2)})
    by_snr = pd.DataFrame(rows)
    print('Clean QTDB:', clean_metrics)
    display(by_snr.round(3))
    plt.plot(by_snr.input_snr_db, by_snr.output_snr_db, 'o-')
    plt.plot(by_snr.input_snr_db, by_snr.input_snr_db, '--', label='no change')
    plt.xlabel('Input SNR (dB)'); plt.ylabel('Output SNR (dB)'); plt.legend(); plt.show()
    assert (by_snr.mse_after < by_snr.mse_before).all(), 'QTDB recovery made at least one SNR worse'
    return clean_metrics, by_snr

qtdb_clean_metrics, qtdb_by_snr = qtdb_validation()
```

## CSV inference

The next cell accepts a single-lead CSV, filters it, uses overlapping windows, averages the predicted artifacts, and writes a recovered signal. Pass the actual sampling rate; non-360 Hz signals are resampled internally and restored to their original length.

```python
def read_single_lead_csv(csv_path, column=None):
    frame = pd.read_csv(csv_path)
    if column is None:
        numeric = frame.select_dtypes(include=np.number).columns
        if len(numeric) == 0:
            raise ValueError('CSV must contain a numeric ECG column; pass column=... if needed.')
        column = numeric[0]
    values = pd.to_numeric(frame[column], errors='coerce').dropna().to_numpy(dtype=np.float64)
    if len(values) < WINDOW:
        raise ValueError(f'ECG needs at least {WINDOW} samples after removing missing values.')
    return values, column

def recover_ecg(signal, input_fs=FS, batch_size=256):
    original = np.asarray(signal, dtype=np.float64)
    if input_fs != FS:
        ratio = Fraction(FS / input_fs).limit_denominator(1000)
        work = resample_poly(original, ratio.numerator, ratio.denominator)
    else:
        work = original.copy()
    filtered = bandpass_filter(work)
    starts = list(range(0, max(1, len(filtered) - WINDOW + 1), STRIDE))
    if starts[-1] + WINDOW < len(filtered): starts.append(len(filtered) - WINDOW)
    artifact_sum, weights = np.zeros_like(filtered), np.zeros_like(filtered)
    model.eval()
    with torch.no_grad():
        for first in range(0, len(starts), batch_size):
            batch_starts = starts[first:first + batch_size]
            batch = np.stack([filtered[s:s + WINDOW] for s in batch_starts]) / NORMALIZATION_SCALE
            predicted = model(torch.from_numpy(batch.astype(np.float32)).unsqueeze(1).to(device)).cpu().numpy()[:, 0]
            for start, artifact in zip(batch_starts, predicted * NORMALIZATION_SCALE):
                artifact_sum[start:start + WINDOW] += artifact
                weights[start:start + WINDOW] += 1
    recovered = filtered - artifact_sum / np.maximum(weights, 1)
    return resample(recovered, len(original)) if input_fs != FS else recovered

# Example (uncomment and set the sampling rate/column for your file):
raw_ecg, used_column = read_single_lead_csv('ecg.csv', column=None)
recovered_ecg = recover_ecg(raw_ecg, input_fs=360)
pd.DataFrame({'recovered_ecg': recovered_ecg}).to_csv('recovered_ecg.csv', index=False)
print(f'Processed {used_column} -> recovered_ecg.csv')
```

## QTDB pass rule

Do not use held-out MIT-BIH score as final claim. Use the QTDB cell above. Keep model only when clean QTDB artifact RMS is tiny, clean morphology correlation stays near 1, and every tested noisy SNR has lower MSE after recovery.

```python
def plot_qtdb_examples(model, n_examples=5):
    model.eval()
    rng = np.random.default_rng(SEED + 123)

    fig, axes = plt.subplots(n_examples, 1, figsize=(15, 3*n_examples))

    if n_examples == 1:
        axes = [axes]

    for ax in axes:
        # Pick random QTDB record
        record_name = rng.choice(QTDB_RECORDS)
        record = wfdb.rdrecord(str(DATA_ROOT / "qtdb" / record_name))

        signal = np.asarray(record.p_signal[:, 0], dtype=np.float64)

        if round(record.fs) != FS:
            signal = resample_poly(signal, FS, round(record.fs))

        start = rng.integers(CONTEXT, len(signal) - WINDOW - CONTEXT)

        clean_segment = signal[start-CONTEXT:start+WINDOW+CONTEXT]
        clean = bandpass_filter(clean_segment)[CONTEXT:CONTEXT+WINDOW]

        # Random noise
        requested_snr = rng.choice([-6,-3,0,3,6,9,12,15,20])

        noise = fresh_noise[
            NOISE_RECORDS[int(rng.integers(len(NOISE_RECORDS)))]
        ]

        noise_start = rng.integers(
            0,
            len(noise) - len(clean_segment)
        )

        artifact = noise[
            noise_start:noise_start+len(clean_segment)
        ]

        artifact *= (
            np.sqrt(np.mean(clean_segment**2))
            /
            (
                np.sqrt(np.mean(artifact**2))
                * 10**(requested_snr/20)
            )
        )

        noisy = bandpass_filter(
            clean_segment + artifact
        )[CONTEXT:CONTEXT+WINDOW]

        # Ground truth artifact after filtering
        true_artifact = noisy - clean

        # Prediction
        x = (
            torch.tensor(noisy / NORMALIZATION_SCALE,
                         dtype=torch.float32)
            .unsqueeze(0)
            .unsqueeze(0)
            .to(device)
        )

        with torch.no_grad():
            pred_artifact = (
                model(x)
                .cpu()
                .numpy()[0,0]
                * NORMALIZATION_SCALE
            )

        recovered = noisy - pred_artifact

        ax.plot(clean, label="Clean ECG", linewidth=2)
        ax.plot(noisy, label="Noisy ECG", alpha=0.6)
        ax.plot(recovered, label="Recovered ECG", linewidth=2)
        ax.plot(pred_artifact, label="Pred Artifact", linestyle="--")
        ax.plot(true_artifact, label="True Artifact", linestyle=":")

        rmse = np.sqrt(np.mean((recovered-clean)**2))

        ax.set_title(
            f"{record_name} | SNR={requested_snr} dB | RMSE={rmse:.3f}"
        )

        ax.legend(loc="upper right", fontsize=8)
        ax.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
plot_qtdb_examples(model, n_examples=5)
```
